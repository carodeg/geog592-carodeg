definitions.
- **Class** : a template that defines the kind of data and behavior needed to create an *object*.
- **Object** :  one actual instance created from a class.  Objects in a class have *attributes* and *methods.*
- **Attribute** : refers to specific data that belongs to an object. 
- **Method** : refers to a function that belongs to the object; usually all objects of the same class have the same methods (but the methods' behaviour can depent on the specific attributes)
