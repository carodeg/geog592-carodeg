```python
cleaned_data = []

with open(filename, "r", encoding="utf-8-sig") as file:
    reader = csv.DictReader(file)

    next(reader) #Skip second header row

    for row in reader:

        State_fips = row["State FIPS Code"].strip()
        Premature_deathb = row["Premature Death (Black)"].strip()
        Premature_deathw = row["Premature Death (White)"].strip()
        fips = row["5-digit FIPS Code"].strip()

   # skip rows with no FIPS code
        if fips == "":
            continue

        # make sure FIPS stays 5 digits
        fips = fips.zfill(5)

        # check premature death values
        if Premature_deathb == "":
            Premature_deathb = None
        else:
            try:
                Premature_deathb = float(Premature_deathb)
            except ValueError:
                Premature_deathb = None

        if Premature_deathw == "":
            Premature_deathw = None
        else:
            try:
                Premature_deathw = float(Premature_deathw)
            except ValueError:
                Premature_deathw = None

        cleaned_data.append({
            "State FIPS Code": State_fips,
            "Premature Death Black": Premature_deathb,
            "Premature Death White": Premature_deathw,
            "FIPS Code": fips
        })

print("Number of cleaned rows:", len(cleaned_data))
print(cleaned_data[:5])
```

```python
output_file = "cleanedanalytic_data2025_v3.csv"

fieldnames = [
    "State FIPS Code",
    "Premature Death Black",
    "Premature Death White",
    "FIPS Code"
]

with open(output_file, "w", newline="", encoding="utf-8") as file:
    writer = csv.DictWriter(file, fieldnames=fieldnames)

    writer.writeheader()
    writer.writerows(cleaned_data)

print("Cleaned CSV created!", output_file)
```
