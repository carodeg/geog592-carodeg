1. Reading original csv file:

```python
import csv

filename = "tabela1613.csv"

with open(filename, "r", encoding="utf-8-sig") as file:
    reader = csv.reader(file)
    rows = list(reader)

for row in rows[:10]:
    print(row)    
```

3. Filter the data so that only municipalities in Pará (PA) remain.

```python
para_rows = []

for row in rows[5:]:
    if len(row) > 5 and row[2].endswith("(PA)"):
        para_rows.append(row)

print(len(para_rows))

for row in para_rows[:10]:
    print(row)
```

4. Keep only the columns containing:
the municipality code (Cod / Cód.), which will later be used to join the CSV to a GIS layer;
the quantity of açaí produced.

```python
selected_rows = []

for row in para_rows:
    code = row[1]
    acai = row[5]

    selected_rows.append([code, acai])

for row in selected_rows[:10]:
    print(row)
```

6. Check the selected data for problems such as:
missing values;
non-numeric values in the açaí production column;
unnecessary rows or columns.
data type

```python
for row in selected_rows:
    code = row[0]
    acai = row[1]

    try:
        value = float(acai)
    except ValueError:
        print("Non-numeric value:", code, acai)
```

```python
cleaned_rows = []

for row in selected_rows:
    code = row[0]
    acai = row[1]

    if acai == "-":
        acai = 0

    elif acai == "..." or acai == ".." or acai == "":
        acai = ""

    else:
        acai = float(acai)

    cleaned_rows.append([code, acai])
```

```python
for row in cleaned_rows[:10]:
    print(row)
```

```python
print("Number of Pará municipalities:", len(cleaned_rows))
```

```python
codes = []

for row in cleaned_rows:
    codes.append(row[0])

print("Unique codes:", len(set(codes)))
print("Total rows:", len(codes))
```

```python
output_filename = "para_acai_production.csv"

with open(output_filename, "w", newline="", encoding="utf-8-sig") as file:
    writer = csv.writer(file)

    writer.writerow(["cod", "acai_production"])
    writer.writerows(cleaned_rows)

print("File saved:", output_filename)
```
